Practice Pattern Problem


```python
n = 5
for i in range(1, n+1):
  for j in range(1, i+1):
    print("*", end = " ")
  print()
```

```python
for i in range(n, 0, -1):
  for j in range(1, i+1):
    print("*", end = " ")
  print()
```

```python
n = 5
for i in range(1, n+1):
    print(' ' * (n-i), end = "")
    print("*" * (2*i-1))
```
